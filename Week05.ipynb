```python
import cv2 as cv
from google.colab.patches import cv2_imshow
from skimage import io
import numpy as np
import matplotlib.pyplot as plt
import math
import os
import glob
```

```python
# 1. Membaca gambar dan konversi ruang warna
img = cv.imread('/content/drive/MyDrive/Colab Notebooks/PCVK_2024/Image Jobsheet/lena.jpg')
img = cv.cvtColor(img, cv.COLOR_BGR2RGB)

# 2. Mengambil dimensi gambar dan menyiapkan penampung frekuensi
height, width, depth = np.shape(img)
names = np.arange(256)

red = [0] * 256
green = [0] * 256
blue = [0] * 256

# 3. Menghitung frekuensi kemunculan setiap intensitas piksel secara manual
for y in range(0, height):
    for x in range(0, width):
        red[img[y][x][0]] += 1
        green[img[y][x][1]] += 1
        blue[img[y][x][2]] += 1

# 4. Menampilkan histogram menggunakan Matplotlib
names = np.arange(256)
fig, axs = plt.subplots(1, 3, figsize=[20, 5], sharex=True, sharey=True)
fig.suptitle('Histogram RGB plot')
fig.text(0.09, 0.5, 'Jumlah Kemunculan', va='center', rotation='vertical')
fig.text(0.5, 0.04, 'Intensitas Warna', ha='center')

axs[0].bar(names, red, color='red')
axs[1].bar(names, green, color='green')
axs[2].bar(names, blue, color='blue')

plt.show()
```
