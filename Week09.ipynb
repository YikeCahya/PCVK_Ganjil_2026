Fungsi dilasi serta tampilkan hasil dilasi pada citra dengan Structuring Element
Dilasi (SED) berbentuk cross 3 x 3 tanpa menggunakan library morphology dari
openCV

```python
#Dilasi tanpa library
def dilasi_citra(F, w):
    # size image
    p, q = F.shape
    imgD = np.zeros((p, q), dtype=np.uint8)
    # Generate structure element dilasi (SED)
    SED = np.ones((w, w), dtype=np.uint8)
    constant1 = (w - 1) // 2
    # Proses Dilasi
    for i in range(constant1, p - constant1):
        for j in range(constant1, q - constant1):
            temp = F[i - constant1 : i + constant1 + 1, j - constant1 : j + constant1 + 1]
            product = temp * SED
            imgD[i, j] = np.max(product)
    return imgD

img_d = cv2.imread('/content/drive/MyDrive/Kuliah/PCVK/Images/plat_nomor.jpg', 0)
ret, thresh = cv2.threshold(img_d, 125, 200, cv2.THRESH_BINARY)
w = 5
imgD = dilasi_citra(thresh, w)

plt.subplot(131)
plt.imshow(img_d, cmap='gray')
plt.title('Citra Awal')
plt.xticks([])
plt.yticks([])

plt.subplot(132)
plt.imshow(imgD, cmap='gray')
plt.title('Citra Hasil Dilasi')
plt.xticks([])
plt.yticks([])

plt.show()
```

Fungsi erosi serta tampilkan hasil erosi pada citra dengan Structuring Element
Erosi (SE) berbentuk square 5 x 5 tanpa menggunakan library morphology dari
openCV.

```python
#Erosi Tanpa Library
def erosi_citra(F, k):
    # size image
    m, n = F.shape
    # Generate structure element erosi (SE)
    SE = np.ones((k, k), dtype=np.uint8)
    constant = (k - 1) // 2
    imgE = np.zeros((m, n), dtype=np.uint8)
    # Proses Erosi
    for i in range(constant, m - constant):
        for j in range(constant, n - constant):
            temp = F[i - constant : i + constant + 1, j - constant : j + constant + 1]
            product = temp * SE
            imgE[i, j] = np.min(product)
    return imgE

img_e = cv2.imread('/content/drive/MyDrive/Kuliah/PCVK/Images/plat_nomor.jpg', 0)
ret1, thresh1 = cv2.threshold(img_e, 175, 225, cv2.THRESH_BINARY)
k = 5
imgE = erosi_citra(thresh1, k)

plt.subplot(131)
plt.imshow(img_e, cmap='gray')
plt.title('Citra Awal')
plt.xticks([])
plt.yticks([])

plt.subplot(132)
plt.imshow(imgE, cmap='gray')
plt.title('Citra Hasil Erosi')
plt.xticks([])
plt.yticks([])

plt.show()
```

```python
from google.colab import drive

# Tesseract OCR (untuk membuktikan citra "dapat dibaca")
!apt-get -qq install tesseract-ocr tesseract-ocr-ind > /dev/null
!pip -q install pytesseract

import cv2
import numpy as np
import pytesseract
from matplotlib import pyplot as plt

KTP_PATH = '/content/drive/MyDrive/Kuliah/PCVK/Images/KTP IKE ai.jpg'
OUT_PATH = '/content/drive/MyDrive/PCVK/Images/ktp_morfologi.png'

KEYWORDS = ['NIK', 'NAMA', 'TEMPAT', 'LAHIR', 'KELAMIN', 'ALAMAT', 'RT/RW', 'KEL', 'KECAMATAN',
            'AGAMA', 'STATUS', 'PERKAWINAN', 'PEKERJAAN', 'KEWARGANEGARAAN', 'BERLAKU']

# ---------- Fungsi ----------
def load_ktp(path, width=1200):
    img = cv2.imread(path)
    if img is None:
        raise FileNotFoundError(f'Citra tidak ditemukan: {path}')
    h, w = img.shape[:2]
    img = cv2.resize(img, (width, int(h * width / w)), interpolation=cv2.INTER_CUBIC)
    return img, cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

def rect(size):
    return cv2.getStructuringElement(cv2.MORPH_RECT, size)

def pipeline_morfologi(gray, bh=(15, 15), op=(2, 2), cl=(3, 3), dl=(2, 2)):
    blur = cv2.GaussianBlur(gray, (3, 3), 0)
    blackhat = cv2.morphologyEx(blur, cv2.MORPH_BLACKHAT, rect(bh)) # ekstrak teks gelap
    _, biner = cv2.threshold(blackhat, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    opened = cv2.morphologyEx(biner, cv2.MORPH_OPEN, rect(op))    # hapus noise
    closed = cv2.morphologyEx(opened, cv2.MORPH_CLOSE, rect(cl))  # sambung goresan putus
    final = cv2.dilate(closed, rect(dl), iterations=1)            # tebalkan goresan
    return {'Black-Hat': blackhat, 'Biner (Otsu)': biner, 'Opening': opened,
            'Closing': closed, 'Dilasi (final)': final}

def ocr(citra, invert=False):
    if invert:
        citra = 255 - citra
    try:
        return pytesseract.image_to_string(citra, lang='ind+eng', config='--psm 6')
    except pytesseract.TesseractError:
        return pytesseract.image_to_string(citra, lang='eng', config='--psm 6')

def skor(teks):
    t = teks.upper()
    return sum(k in t for k in KEYWORDS)

# ---------- 1. Load & grayscale ----------
img, gray = load_ktp(KTP_PATH)
plt.figure(figsize=(12, 4))
plt.subplot(121), plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)), plt.title('Citra KTP Asli'), plt.xticks([]), plt.yticks([])
plt.subplot(122), plt.imshow(gray, cmap='gray'), plt.title('Grayscale'), plt.xticks([]), plt.yticks([])
plt.show()

# ---------- 2. Baseline (tanpa morfologi) ----------
_, baseline = cv2.threshold(cv2.GaussianBlur(gray, (3, 3), 0), 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

# ---------- 3. Pipeline morfologi ----------
hasil = pipeline_morfologi(gray)
plt.figure(figsize=(14, 8))
for i, (nama, citra) in enumerate(hasil.items(), 1):
    plt.subplot(3, 2, i), plt.imshow(citra, cmap='gray'), plt.title(nama), plt.xticks([]), plt.yticks([])
plt.tight_layout(), plt.show()

# ---------- 4. Evaluasi OCR ----------
teks_asli     = ocr(gray)
teks_baseline = ocr(baseline)
teks_morfo    = ocr(hasil['Dilasi (final)'], invert=True)
print(f'Skor keyword -> Grayscale: {skor(teks_asli)} | Baseline Otsu: {skor(teks_baseline)} | Morfologi: {skor(teks_morfo)}  (maks {len(KEYWORDS)})')

# ---------- 5. Eksperimen parameter ----------
percobaan = []
for bh in [(11, 11), (15, 15), (21, 21)]:
    for cl in [(1, 1), (2, 2), (3, 3)]:
        s = skor(ocr(pipeline_morfologi(gray, bh=bh, cl=cl)['Dilasi (final)'], invert=True))
        percobaan.append((s, bh, cl))
        print(f'Black-Hat SE={bh} | Closing SE={cl} -> skor {s}')
terbaik = max(percobaan, key=lambda x: x[0])
print('\nKonfigurasi terbaik (skor, Black-Hat, Closing):', terbaik)

# ---------- 6. Hasil akhir ----------
_, bh_best, cl_best = terbaik
final_best = pipeline_morfologi(gray, bh=bh_best, cl=cl_best)['Dilasi (final)']
plt.figure(figsize=(14, 4))
plt.subplot(131), plt.imshow(gray, cmap='gray'), plt.title('Citra Awal (Grayscale)'), plt.xticks([]), plt.yticks([])
plt.subplot(132), plt.imshow(baseline, cmap='gray'), plt.title('Baseline Otsu'), plt.xticks([]), plt.yticks([])
plt.subplot(133), plt.imshow(255 - final_best, cmap='gray'), plt.title(f'Morfologi BH{bh_best} CL{cl_best}'), plt.xticks([]), plt.yticks([])
plt.tight_layout(), plt.show()

cv2.imwrite(OUT_PATH, 255 - final_best)
print('\n===== HASIL OCR (Morfologi terbaik) =====')
print(ocr(final_best, invert=True))
```
